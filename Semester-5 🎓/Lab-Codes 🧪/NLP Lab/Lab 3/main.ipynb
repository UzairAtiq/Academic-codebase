# Augmentation
We generate 5 different variants of a sentence using basic data augmentation techniques.

```python
# Task 1: Generate multiple augmented variants of a sentence
import random

# Sample input sentence
sentence = "The movie was really excellent."

# Function to randomly apply one augmentation technique
def augment(text):
    words = text.split()
    technique = random.choice(["synonym", "delete", "insert", "swap"])
    
    # Synonym replacement
    if technique == "synonym":
        synonyms = {"movie": "film", "excellent": "great", "really": "very"}
        words = [synonyms.get(w.lower().strip("."), w) for w in words]
    # Random deletion
    elif technique == "delete" and len(words) > 2:
        words.pop(random.randint(0, len(words) - 1))
    # Random insertion
    elif technique == "insert":
        words.insert(random.randint(0, len(words)), "very")
    # Random swap
    elif technique == "swap" and len(words) >= 2:
        i, j = random.sample(range(len(words)), 2)
        words[i], words[j] = words[j], words[i]
        
    return " ".join(words)

# Display original sentence
print("Original:", sentence)
print("\nVariants:")

# Generate 5 augmented variants
for i in range(1, 6):
    print(f"Variant {i}:", augment(sentence))

# Quality check on output quality
print("\nQuality Check:")
print("- Swap and delete can make the sentence awkward or ungrammatical.")
print("- Synonym and insert preserve the sentence meaning well.")
```

# Program
We create a sentiment dataset, augment each sentence 3 times, and combine them into a larger dataset.

```python
# Task 2: Complete sentiment dataset augmentation program
import pandas as pd
import random

# 1. Create a simple labeled sentiment dataset
data = [
    {"text": "The movie was great", "label": "positive"},
    {"text": "I loved this film", "label": "positive"},
    {"text": "The food was terrible", "label": "negative"},
    {"text": "I hated the service", "label": "negative"}
]
df_original = pd.DataFrame(data)

# 2. Define a simple augmentation function using synonyms and insertion
synonyms = {
    "great": "amazing", "loved": "liked", "movie": "film",
    "terrible": "awful", "hated": "disliked", "food": "meal"
}

def augment(text):
    words = text.split()
    if random.choice(["synonym", "insert"]) == "synonym":
        words = [synonyms.get(w.lower(), w) for w in words]
    else:
        words.insert(random.randint(0, len(words)), "really")
    return " ".join(words)

# 3. Generate 3 augmented examples for each original row
augmented_data = []
for row in data:
    for _ in range(3):
        augmented_data.append({"text": augment(row["text"]), "label": row["label"]})

# 4. Combine original and augmented data into a single DataFrame
df_augmented = pd.DataFrame(augmented_data)
df_final = pd.concat([df_original, df_augmented], ignore_index=True)

# 5. Display original and final datasets
print("Original Dataset (Size:", len(df_original), "):")
print(df_original)
print("\nFinal Combined Dataset (Size:", len(df_final), "):")
print(df_final)
```

# Exercises
We complete student exercises on synonym replacement, deletion rates, insertion, swapping, dataset balancing, and classification.

```python
# Task 3: Student Exercises
import pandas as pd
import random
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report

# 1 & 2. 10 sentiment sentences and synonym replacement
synonyms = {"good": "great", "bad": "terrible", "loved": "liked", "hated": "disliked", "happy": "pleased"}
sentences = [
    ("The movie was good", "positive"),
    ("I loved the story", "positive"),
    ("The acting was great", "positive"),
    ("I am happy today", "positive"),
    ("Everything was wonderful", "positive"),
    ("The movie was bad", "negative"),
    ("I hated the story", "negative"),
    ("The acting was terrible", "negative"),
    ("I am sad today", "negative"),
    ("Everything was awful", "negative")
]

# Generate 2 augmented examples per sentence
augmented_list = []
for text, label in sentences:
    words = [synonyms.get(w.lower(), w) for w in text.split()]
    augmented_list.append((" ".join(words), label))
    augmented_list.append((text + " really", label))

print(f"1 & 2. Generated {len(augmented_list)} augmented examples from 10 sentences.")

# 3. Test deletion with probabilities 0.1, 0.2, 0.3
sample = "The movie was really very good and entertaining"
words = sample.split()
random.seed(42)
for p in [0.1, 0.2, 0.3]:
    kept = [w for w in words if random.random() > p]
    print(f"3. Deletion (prob={p}):", " ".join(kept))

# 4. Insertion and Swapping
w_insert = words.copy()
w_insert.insert(2, "very")
print("4. Insertion:", " ".join(w_insert))

w_swap = words.copy()
w_swap[0], w_swap[1] = w_swap[1], w_swap[0]
print("4. Swapping :", " ".join(w_swap))

# 5. Five rejected sentences
rejected = [
    ("not good -> good", "Negation removed, flips label to positive"),
    ("movie the was good", "Grammar broken by random swapping"),
    ("The", "Too short due to high deletion"),
    ("good -> bad", "Antonym replaced instead of synonym"),
    ("good good good movie", "Repeated insertion creates unnatural text")
]
print("\n5. Five Sentences to Reject:")
for s, reason in rejected:
    print(f" - {s}: {reason}")

# 6, 7 & 8. Split imbalanced data and balance minority class
df = pd.DataFrame([
    ("Great movie", 1), ("Loved it", 1), ("Very nice", 1), ("Enjoyed it", 1),
    ("Superb film", 1), ("Awesome show", 1), ("Bad movie", 0), ("Terrible show", 0)
], columns=["text", "label"])

train_df, test_df = train_test_split(df, test_size=0.25, random_state=42)
print("\n8. Before balancing Train:\n", train_df["label"].value_counts())

minority = train_df[train_df["label"] == 0]
extra_rows = []
for _, row in minority.iterrows():
    extra_rows.append({"text": row["text"] + " really", "label": 0})
    extra_rows.append({"text": row["text"].replace("Bad", "Awful"), "label": 0})

train_balanced = pd.concat([train_df, pd.DataFrame(extra_rows)], ignore_index=True)
print("\n8. After balancing Train:\n", train_balanced["label"].value_counts())

# 9 & 10. Train and evaluate classifier
cv = CountVectorizer()
X_train_orig = cv.fit_transform(train_df["text"])
X_train_aug = cv.transform(train_balanced["text"])
X_test = cv.transform(test_df["text"])

clf_orig = MultinomialNB().fit(X_train_orig, train_df["label"])
print("\n10. Baseline Report:")
print(classification_report(test_df["label"], clf_orig.predict(X_test), zero_division=0))

clf_aug = MultinomialNB().fit(X_train_aug, train_balanced["label"])
print("10. Augmented Report:")
print(classification_report(test_df["label"], clf_aug.predict(X_test), zero_division=0))
```

# Challenge
We implement the sentiment augmentation challenge by augmenting training data with three techniques and evaluating model accuracy.

```python
# Task 4: Sentiment Data Augmentation Challenge
import pandas as pd
import random
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# 11. Create a labeled sentiment dataset
dataset = [
    ("This film was wonderful and brilliant", 1),
    ("I truly enjoyed this amazing movie", 1),
    ("Great acting and great storyline", 1),
    ("Superb performance by the entire cast", 1),
    ("A fantastic movie worth watching", 1),
    ("Loved the cinematography and music", 1),
    ("The plot was completely terrible and boring", 0),
    ("I hated this horrible film", 0),
    ("Awful acting and terrible direction", 0),
    ("Worst movie experience of my life", 0),
    ("The script was bad and dull", 0),
    ("Waste of time and very disappointing", 0)
]
df = pd.DataFrame(dataset, columns=["text", "label"])

# 12. Split train and test before augmentation to avoid data leakage
train_df, test_df = train_test_split(df, test_size=0.25, random_state=42, stratify=df["label"])
print("12. Train size:", len(train_df), "| Test size:", len(test_df))

# 13, 14 & 15. Augment only training data using 3 techniques preserving labels
synonyms = {"wonderful": "great", "enjoyed": "liked", "great": "good", "terrible": "awful", "hated": "disliked", "bad": "poor"}

def augment_text(text):
    words = text.split()
    tech = random.choice(["synonym", "insert", "delete"])
    if tech == "synonym":
        words = [synonyms.get(w.lower(), w) for w in words]
    elif tech == "insert":
        words.insert(0, "really")
    elif tech == "delete" and len(words) > 3:
        words.pop(random.randint(0, len(words) - 1))
    return " ".join(words)

# Generate 3 augmented rows per training row
aug_list = []
for _, row in train_df.iterrows():
    for _ in range(3):
        aug_list.append({"text": augment_text(row["text"]), "label": row["label"]})

# 16 & 17. Quality control and duplicate removal
train_augmented = pd.concat([train_df, pd.DataFrame(aug_list)], ignore_index=True)
train_augmented = train_augmented.drop_duplicates(subset=["text"]).reset_index(drop=True)

# Clearly display original and augmented datasets
print("\nOriginal Training Dataset:")
print(train_df)
print("\nAugmented Training Dataset (Sample):")
print(train_augmented.head(10))

# 18. Compare class distributions
print("\n18. Class Distribution Before:\n", train_df["label"].value_counts())
print("\n18. Class Distribution After:\n", train_augmented["label"].value_counts())

# 19 & 20. Train baseline and augmented models
cv = CountVectorizer()
X_tr_base = cv.fit_transform(train_df["text"])
X_tr_aug = cv.transform(train_augmented["text"])
X_te = cv.transform(test_df["text"])

model_base = MultinomialNB().fit(X_tr_base, train_df["label"])
model_aug = MultinomialNB().fit(X_tr_aug, train_augmented["label"])

pred_base = model_base.predict(X_te)
pred_aug = model_aug.predict(X_te)

# 21. Baseline vs augmented evaluation table
y_true = test_df["label"]
eval_table = pd.DataFrame({
    "Metric": ["Accuracy", "Precision", "Recall", "F1-Score"],
    "Baseline Model": [
        f"{accuracy_score(y_true, pred_base):.2f}",
        f"{precision_score(y_true, pred_base, zero_division=0):.2f}",
        f"{recall_score(y_true, pred_base, zero_division=0):.2f}",
        f"{f1_score(y_true, pred_base, zero_division=0):.2f}"
    ],
    "Augmented Model": [
        f"{accuracy_score(y_true, pred_aug):.2f}",
        f"{precision_score(y_true, pred_aug, zero_division=0):.2f}",
        f"{recall_score(y_true, pred_aug, zero_division=0):.2f}",
        f"{f1_score(y_true, pred_aug, zero_division=0):.2f}"
    ]
})

print("\n21. Baseline vs Augmented Evaluation Table:")
print(eval_table.to_string(index=False))

# 22. Conclusion on data quality, label preservation, and model impact
print("\n22. Conclusion:")
print("1. Data Quality: Removing duplicates and broken phrases ensures high training data quality.")
print("2. Label Preservation: Careful synonym and insertion techniques preserve sentiment polarity without label drift.")
print("3. Model Impact: Augmentation expands vocabulary diversity and helps small-data models generalize to new phrasing.")
```
