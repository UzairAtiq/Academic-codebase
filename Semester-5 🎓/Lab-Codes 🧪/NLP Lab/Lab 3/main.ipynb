# Dataset
We create a simple text dataset with sentences and their sentiment labels.

```python
# Create and display a simple sentiment dataset using pandas
import pandas as pd

data = {
    "text": [
        "I love this movie.",
        "The movie was excellent.",
        "This film was amazing.",
        "I hate this movie.",
        "The movie was terrible.",
        "This film was boring."
    ],
    "label": [
        "positive", "positive", "positive",
        "negative", "negative", "negative"
    ]
}

df = pd.DataFrame(data)
print(df)
```

# Replacement
We replace selected words with their synonyms from a dictionary.

```python
# Replace words with synonyms using a dictionary
import random

synonyms = {
    "good": ["great", "excellent", "nice"],
    "excellent": ["great", "fantastic"],
    "amazing": ["wonderful", "fantastic"],
    "love": ["like", "enjoy"],
    "terrible": ["awful", "horrible"],
    "hate": ["dislike"],
    "boring": ["dull", "uninteresting"]
}

def synonym_replace(sentence):
    words = sentence.split()
    result = []
    for word in words:
        clean = word.lower().strip(".,!?")
        if clean in synonyms:
            result.append(random.choice(synonyms[clean]))
        else:
            result.append(word)
    return " ".join(result)

text = "The movie was excellent."
print("Original:", text)
print("Augmented:", synonym_replace(text))
```
