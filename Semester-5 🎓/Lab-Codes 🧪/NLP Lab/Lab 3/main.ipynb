# Augmentation
We randomly apply synonym replacement, deletion, insertion, or swapping to create sentence variants.

```python
# Task 1: Generate multiple augmented examples using 4 techniques
import random

# Synonyms dictionary for replacement
synonyms = {
    "movie": ["film", "show"],
    "excellent": ["great", "fantastic"],
    "really": ["truly", "very"],
    "good": ["great", "nice"]
}

# 1. Synonym replacement
def synonym_replace(sentence):
    words = sentence.split()
    new_words = []
    for word in words:
        clean = word.lower().strip(".,!?")
        if clean in synonyms:
            new_words.append(random.choice(synonyms[clean]))
        else:
            new_words.append(word)
    return " ".join(new_words)

# 2. Random deletion
def random_deletion(sentence, prob=0.2):
    words = sentence.split()
    if len(words) <= 2:
        return sentence
    kept = [w for w in words if random.random() > prob]
    if not kept:
        kept = [random.choice(words)]
    return " ".join(kept)

# 3. Random insertion
def random_insertion(sentence, insert_word="really"):
    words = sentence.split()
    pos = random.randint(0, len(words))
    words.insert(pos, insert_word)
    return " ".join(words)

# 4. Random swap
def random_swap(sentence):
    words = sentence.split()
    if len(words) < 2:
        return sentence
    i, j = random.sample(range(len(words)), 2)
    words[i], words[j] = words[j], words[i]
    return " ".join(words)

# Combined function to randomly pick one augmentation technique
def augment(sentence):
    technique = random.choice([synonym_replace, random_deletion, random_insertion, random_swap])
    return technique(sentence)

# Original sentence
sentence = "The movie was really excellent."
print("Original Sentence:", sentence)
print("\nGenerated Variants:")

# Generate 5 variants of the sentence
for i in range(1, 6):
    variant = augment(sentence)
    print(f"Variant {i}: {variant}")

# Quality check on generated variants
print("\nQuality Check:")
print("- Awkward output: Swapping words or deleting key nouns can break grammar.")
print("- Acceptable output: Synonym replacement and insertion preserve meaning well.")
```
