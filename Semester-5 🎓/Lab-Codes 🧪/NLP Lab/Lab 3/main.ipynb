# Dataset
We create a simple text dataset with sentences and their sentiment labels.

```python
# Create and display a simple sentiment dataset using pandas
import pandas as pd

data = {
    "text": [
        "I love this movie.",
        "The movie was excellent.",
        "This film was amazing.",
        "I hate this movie.",
        "The movie was terrible.",
        "This film was boring."
    ],
    "label": [
        "positive", "positive", "positive",
        "negative", "negative", "negative"
    ]
}

df = pd.DataFrame(data)
print(df)
```
