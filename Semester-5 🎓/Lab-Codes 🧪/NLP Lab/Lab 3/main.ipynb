# Augmentation
We generate 5 different variants of a sentence using basic data augmentation techniques.

```python
# Task 1: Generate multiple augmented variants of a sentence
import random

# Sample input sentence
sentence = "The movie was really excellent."

# Function to randomly apply one augmentation technique
def augment(text):
    words = text.split()
    technique = random.choice(["synonym", "delete", "insert", "swap"])
    
    # Synonym replacement
    if technique == "synonym":
        synonyms = {"movie": "film", "excellent": "great", "really": "very"}
        words = [synonyms.get(w.lower().strip("."), w) for w in words]
    # Random deletion
    elif technique == "delete" and len(words) > 2:
        words.pop(random.randint(0, len(words) - 1))
    # Random insertion
    elif technique == "insert":
        words.insert(random.randint(0, len(words)), "very")
    # Random swap
    elif technique == "swap" and len(words) >= 2:
        i, j = random.sample(range(len(words)), 2)
        words[i], words[j] = words[j], words[i]
        
    return " ".join(words)

# Display original sentence
print("Original:", sentence)
print("\nVariants:")

# Generate 5 augmented variants
for i in range(1, 6):
    print(f"Variant {i}:", augment(sentence))

# Quality check on output quality
print("\nQuality Check:")
print("- Swap and delete can make the sentence awkward or ungrammatical.")
print("- Synonym and insert preserve the sentence meaning well.")
```

# Program
We create a sentiment dataset, augment each sentence 3 times, and combine them into a larger dataset.

```python
# Task 2: Complete sentiment dataset augmentation program
import pandas as pd
import random

# 1. Create a simple labeled sentiment dataset
data = [
    {"text": "The movie was great", "label": "positive"},
    {"text": "I loved this film", "label": "positive"},
    {"text": "The food was terrible", "label": "negative"},
    {"text": "I hated the service", "label": "negative"}
]
df_original = pd.DataFrame(data)

# 2. Define a simple augmentation function using synonyms and insertion
synonyms = {
    "great": "amazing", "loved": "liked", "movie": "film",
    "terrible": "awful", "hated": "disliked", "food": "meal"
}

def augment(text):
    words = text.split()
    if random.choice(["synonym", "insert"]) == "synonym":
        words = [synonyms.get(w.lower(), w) for w in words]
    else:
        words.insert(random.randint(0, len(words)), "really")
    return " ".join(words)

# 3. Generate 3 augmented examples for each original row
augmented_data = []
for row in data:
    for _ in range(3):
        augmented_data.append({"text": augment(row["text"]), "label": row["label"]})

# 4. Combine original and augmented data into a single DataFrame
df_augmented = pd.DataFrame(augmented_data)
df_final = pd.concat([df_original, df_augmented], ignore_index=True)

# 5. Display original and final datasets
print("Original Dataset (Size:", len(df_original), "):")
print(df_original)
print("\nFinal Combined Dataset (Size:", len(df_final), "):")
print(df_final)
```
