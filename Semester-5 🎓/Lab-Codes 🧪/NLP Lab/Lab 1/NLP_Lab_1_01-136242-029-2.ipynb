## Task 1: Basic String Operations

```python
sentence = input("Enter a sentence: ")
```

```python
# original, stripped, lowercase, uppercase
print("Original:", sentence)
print("Stripped:", sentence.strip())
print("Lowercase:", sentence.lower())
print("Uppercase:", sentence.upper())
```

```python
# character and word counts
char_count = len(sentence)
word_count = len(sentence.split())
print("Character count:", char_count)
print("Word count:", word_count)
```

```python
# first word, last word, first ten characters
words = sentence.split()
print("First word:", words[0])
print("Last word:", words[-1])
print("First 10 characters:", sentence[:10])
```

```python
# count a selected word (case-insensitive)
search_word = input("Enter a word to count: ")
count = sentence.lower().split().count(search_word.lower())
print(f"'{search_word}' appears {count} time(s)")
```

```python
# replace one word and join tokens back into a sentence
old_word = input("Enter the word to replace: ")
new_word = input("Enter the new word: ")
tokens = sentence.split()
updated_tokens = []
for token in tokens:
    if token.lower() == old_word.lower():
        updated_tokens.append(new_word)
    else:
        updated_tokens.append(token)
updated_sentence = " ".join(updated_tokens)
print("Updated sentence:", updated_sentence)
```

## Task 2: Text Cleaning with Regex

```python
import re

text = """
Contact Ali Khan at ali.khan99@aBc123#&$.com or Sara Ahmed at sara_ahmed@yahoo.com.
Ali's phone number is 0312-1234567 and Sara can be reached at 0300 7654321.
Meeting is scheduled on 05-09-2026, follow-up on 12-10-2026.
Please Confirm your Attendance   before the Deadline!!!
"""
```

### Extracting information with regex

```python
# extract email addresses
emails = re.findall(r'[\w.+-]+@[\w-]+\.[\w.-]+', text)
print("Emails found:", emails)
```

```python
# extract digit sequences (phone numbers, parts of dates, etc.)
digit_sequences = re.findall(r'\d+', text)
print("Digit sequences found:", digit_sequences)
```

```python
# extract capitalized words
capitalized_words = re.findall(r'\b[A-Z][a-z]*\b', text)
print("Capitalized words found:", capitalized_words)
```

```python
# extract dates in DD-MM-YYYY format
dates = re.findall(r'\b\d{2}-\d{2}-\d{4}\b', text)
print("Dates found:", dates)
```

### Cleaning the text

```python
# remove emails
cleaned_text = re.sub(r'[\w.+-]+@[\w-]+\.[\w.-]+', '', text)

# remove phone numbers (digit groups with optional - or space in the middle)
cleaned_text = re.sub(r'\d{4}[-\s]\d{6,7}', '', cleaned_text)

# remove remaining punctuation
cleaned_text = re.sub(r'[^\w\s]', '', cleaned_text)

print(cleaned_text)
```

```python
# convert to lowercase and normalize repeated whitespace
cleaned_text = cleaned_text.lower()
cleaned_text = re.sub(r'\s+', ' ', cleaned_text).strip()
print(cleaned_text)
```

### Tokenizing and comparing counts

```python
# tokenize original and cleaned text, compare word counts
original_tokens = text.split()
cleaned_tokens = cleaned_text.split()

print("Word count before cleaning:", len(original_tokens))
print("Word count after cleaning:", len(cleaned_tokens))
print("Cleaned tokens:", cleaned_tokens)
```

### Regex patterns explained

| Pattern | What it matches | Limitation |
|---|---|---|
| `[\w.+-]+@[\w-]+\.[\w.-]+` | Emails: word chars/dots/plus/dash, an `@`, a domain, a dot, then the extension | Doesn't validate real extensions, and can grab a trailing period if the email ends a sentence (seen with `sara_ahmed@yahoo.com.`) |
| `\d+` | One or more digits in a row, used for phone numbers and other digit sequences | Can't tell a phone number apart from a date or any other number |
| `\b[A-Z][a-z]*\b` | Words starting with a capital letter followed by lowercase letters | Also picks up capitalized words that aren't names, like the first word of a sentence |
| `\b\d{2}-\d{2}-\d{4}\b` | Dates written as DD-MM-YYYY | Doesn't check if the day/month values are valid (e.g. `41-13-2026` would still match) |
| `\d{4}[-\s]\d{6,7}` | Phone numbers: a 4-digit group, then a dash or space, then 6-7 more digits | Only works for this specific format, won't catch numbers written differently |
| `[^\w\s]` | Any character that isn't a word character or whitespace, used to strip punctuation | Also removes meaningful punctuation, like apostrophes in "Ali's" |
| `\s+` | One or more whitespace characters in a row, used to collapse repeated spaces | Also collapses newlines/tabs into a single space, losing line formatting |
