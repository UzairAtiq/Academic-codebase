### Regex patterns explained- **`[\w.+-]+@[\w-]+\.[\w.-]+`** — matches emails: some word characters/dots/plus/dash, an `@`, a domain name, a dot, then the extension.  *Limitation:* doesn't validate real domain extensions, and if an email is immediately followed by a period ending the sentence, it can grab that trailing period as part of the domain (seen in the sample text with `sara_ahmed@yahoo.com.`).- **`\d+`** — matches one or more digits in a row, used to pull out phone numbers and any other digit sequences.  *Limitation:* it can't tell a phone number apart from a date or any other number, it just grabs digit groups.- **`\b[A-Z][a-z]*\b`** — matches words that start with a capital letter followed by lowercase letters (word boundaries on both sides).  *Limitation:* it also picks up capitalized words that aren't names, like the first word of a sentence (e.g. "Please", "Meeting").- **`\b\d{2}-\d{2}-\d{4}\b`** — matches dates written as two digits, dash, two digits, dash, four digits (DD-MM-YYYY).  *Limitation:* it doesn't check if the day/month values are actually valid (e.g. `41-13-2026` would still match).- **`\d{4}[-\s]\d{6,7}`** — matches phone numbers made of a 4-digit group, then a dash or space, then 6-7 more digits.  *Limitation:* only works for this specific phone format, won't catch numbers written differently (e.g. with country codes or brackets).- **`[^\w\s]`** — matches any character that is not a word character and not whitespace, used to strip punctuation.  *Limitation:* it also removes punctuation that might be meaningful, like apostrophes in words such as "Ali's".- **`\s+`** — matches one or more whitespace characters in a row, used to collapse repeated spaces into one.  *Limitation:* it also collapses newlines/tabs into a single space, so any intentional line formatting is lost.
