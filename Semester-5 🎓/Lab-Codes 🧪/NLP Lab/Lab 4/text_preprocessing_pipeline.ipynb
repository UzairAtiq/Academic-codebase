# Module 1: Interactive Sentence-Level Text Normalization

Implement a modular Python preprocessing function that accepts raw text input and applies sequential transformations:

* **Lowercasing:** Standardize all alphabetic characters to lowercase.
* **Word Tokenization:** Segment the continuous text stream into discrete token units.
* **Punctuation Stripping:** Filter out non-alphanumeric punctuation symbols.
* **Stopword Removal:** Eliminate highly frequent, non-informative grammatical terms.
* **Porter Stemming:** Heuristically truncate word affixes to their base stems.
* **WordNet Lemmatization:** Contextually map words to their authentic dictionary lemmas.


```python
def process_sentence_pipeline(raw_text: str):
    """
    Executes an end-to-end NLP preprocessing sequence on an input sentence.
    Displays intermediate outputs at each stage and returns the final tokens.
    """
    print(f"\n[Input Text]: {raw_text}\n" + "-" * 60)

    # 1. Lowercase normalization
    lowered = raw_text.lower()
    print(f"1. Lowercased           : {lowered}")

    # 2. Token segmentation
    tokens = word_tokenize(lowered)
    print(f"2. Word Tokens          : {tokens}")

    # 3. Punctuation stripping
    punct_free = [tok for tok in tokens if tok not in string.punctuation]
    print(f"3. Punctuation Removed  : {punct_free}")

    # 4. Stopword filtering
    stop_dict = set(stopwords.words('english'))
    filtered = [tok for tok in punct_free if tok not in stop_dict]
    print(f"4. Stopwords Filtered   : {filtered}")

    # 5. Porter Stemming
    stemmer = PorterStemmer()
    stems = [stemmer.stem(tok) for tok in filtered]
    print(f"5. Porter Stemmed       : {stems}")

    # 6. WordNet Lemmatization
    lemmatizer = WordNetLemmatizer()
    lemmas = [lemmatizer.lemmatize(tok, pos='v') for tok in filtered]
    print(f"6. WordNet Lemmatized   : {lemmas}")

    return lemmas
```

```python
# Accept user input with fallback sample for verification
user_query = input("Enter a sentence to preprocess: ")
if not user_query.strip():
    user_query = "Hello there! Natural language processing is exciting and full of opportunities."

output_tokens = process_sentence_pipeline(user_query)
```

# Module 2: Complete Pipeline Preprocessing of an External Text File

Ingest a structured document of at least 100 words, applying stepwise transformations and reporting diagnostics at each milestone:

* **Milestone 1:** Raw file ingestion and baseline token statistics
* **Milestone 2:** Sentence boundary detection and word tokenization
* **Milestone 3:** Case normalization (Lowercasing)
* **Milestone 4:** Punctuation filtering
* **Milestone 5:** Stopword elimination
* **Milestone 6:** Morphological Stemming
* **Milestone 7:** Morphological Lemmatization
* **Milestone 8:** Synthesis of final cleaned document
* **Milestone 9:** Comparative volume analysis before vs. after cleaning


```python
# Step 1: Read the document and establish baseline word volume
with open("sample.txt", "r", encoding="utf-8") as file_handle:
    raw_document = file_handle.read()

initial_token_list = raw_document.split()
print("=" * 65)
print("1. RAW SOURCE DOCUMENT")
print("=" * 65)
print(raw_document)
print(f"\nInitial Word Count: {len(initial_token_list)}")
```

```python
# Step 2: Sentence segmentation and word tokenization
doc_sentences = sent_tokenize(raw_document)
doc_word_tokens = word_tokenize(raw_document)

print("=" * 65)
print("2. SEGMENTATION & TOKENIZATION")
print("=" * 65)
print(f"Detected Sentences ({len(doc_sentences)}):")
for idx, sent in enumerate(doc_sentences, start=1):
    print(f"  [{idx}] {sent}")

print(f"\nTotal Word Tokens ({len(doc_word_tokens)}):")
print(doc_word_tokens)
```

```python
# Step 3: Case Folding (Lowercasing)
lowered_tokens = [tok.lower() for tok in doc_word_tokens]
print("=" * 65)
print("3. CASE-FOLDED TOKENS")
print("=" * 65)
print(lowered_tokens)
```

```python
# Step 4: Punctuation Filtering
tokens_no_punct = [tok for tok in lowered_tokens if tok not in string.punctuation]
print("=" * 65)
print("4. PUNCTUATION REMOVED")
print("=" * 65)
print(tokens_no_punct)
```

```python
# Step 5: Stopword Elimination
stop_words_set = set(stopwords.words("english"))
tokens_no_stop = [tok for tok in tokens_no_punct if tok not in stop_words_set]
print("=" * 65)
print("5. STOPWORDS ELIMINATED")
print("=" * 65)
print(tokens_no_stop)
```

```python
# Step 6: Porter Stemming Transformation
stemmer_engine = PorterStemmer()
stemmed_output = [stemmer_engine.stem(tok) for tok in tokens_no_stop]
print("=" * 65)
print("6. PORTER STEMMING OUTPUT")
print("=" * 65)
print(stemmed_output)
```

```python
# Step 7: WordNet Lemmatization
wn_lemmatizer = WordNetLemmatizer()
lemmatized_output = [wn_lemmatizer.lemmatize(tok, pos='v') for tok in tokens_no_stop]
print("=" * 65)
print("7. WORDNET LEMMATIZATION OUTPUT")
print("=" * 65)
print(lemmatized_output)
```

```python
# Step 8: Document Reconstruction from Clean Tokens
cleaned_document = " ".join(lemmatized_output)
print("=" * 65)
print("8. NORMALIZED AND CLEANED DOCUMENT")
print("=" * 65)
print(cleaned_document)
```

```python
# Step 9: Quantitative Metric Comparison
original_count = len(initial_token_list)
processed_count = len(cleaned_document.split())
net_reduction = original_count - processed_count
pct_reduction = (net_reduction / original_count) * 100

print("=" * 65)
print("9. PREPROCESSING METRIC SUMMARY")
print("=" * 65)
print(f"Token count PRIOR to preprocessing : {original_count}")
print(f"Token count AFTER preprocessing    : {processed_count}")
print(f"Net tokens eliminated              : {net_reduction} ({pct_reduction:.1f}% reduction)")
```

### Theoretical Evaluation & Conceptual Review Questions

#### 1. What is text preprocessing in NLP?
Text preprocessing refers to the series of programmatic techniques used to purify, normalize, and format unstructured raw natural language text into a uniform, noise-free representation appropriate for downstream machine learning and deep learning architectures.

#### 2. Why is text preprocessing important before training an NLP model?
Natural human text contains high typographical variance, grammatical inflections, irrelevant punctuation, and uninformative filler words. Preprocessing standardizes this heterogeneous vocabulary, mitigates high-dimensional feature sparsity, suppresses ambient noise, and accelerates model convergence.

#### 3. What is tokenization?
Tokenization is the lexical segmentation process of partitioning an unbroken stream of characters into discrete syntactic or semantic elements called **tokens**, which typically correspond to sentences, subwords, or individual words.

#### 4. What is the difference between sentence tokenization and word tokenization?
* **Sentence Tokenization:** Divides a paragraph into distinct sentence segments based on termination boundary markers (periods, exclamation marks, question marks).
* **Word Tokenization:** Further subdivides a sentence into constituent word and symbol units, isolating words from surrounding punctuation.

#### 5. What is lowercasing in text preprocessing?
Lowercasing (also termed case folding) is the normalization process of mapping every alphabetical character across the text corpus to its uniform lowercase representation.

#### 6. Why do we convert text into lowercase?
Computers assess strings strictly by byte values. Without case folding, variants such as `"Language"`, `"LANGUAGE"`, and `"language"` are treated as entirely separate vocabulary entries. Converting everything to lowercase unifies identical semantic words and prevents artificial vocabulary inflation.

#### 7. What is punctuation removal?
Punctuation removal is the filtering stage where syntactic symbols (such as commas, colons, hyphens, and quotation marks) are stripped from the tokenized stream.

#### 8. Why is punctuation removed from text in some NLP tasks?
In tasks like topic classification, spam filtering, and information retrieval (bag-of-words or TF-IDF models), punctuation marks generally contribute negligible domain sentiment or topic signal, while introducing extra tokens that bloat feature vectors.

#### 9. What are stopwords?
Stopwords are high-frequency functional words in a natural language (e.g., pronouns, prepositions, conjunctions, and auxiliary verbs) that provide grammatical cohesion but carry minimal domain-specific lexical content.

#### 10. Give five examples of common English stopwords.
Common English examples include: `about`, `between`, `during`, `under`, `which`.

#### 11. Why are stopwords removed during text preprocessing?
Discarding stopwords eliminates lexical noise, compresses the document size, and directs machine learning algorithms toward the salient, domain-rich content words (nouns, main verbs, descriptive adjectives).

#### 12. Why should words such as “not” sometimes be kept instead of removed?
In sentiment analysis, opinion mining, or logical inference, negation words (`not`, `never`, `hardly`) fundamentally invert the polarity of adjacent words. Removing `"not"` would cause a model to conflate `"not reliable"` with `"reliable"`, leading to catastrophic misclassifications.

#### 13. What is stemming?
Stemming is a fast, rule-driven heuristic technique that strips inflectional prefixes and suffixes from words to reduce them to a shared base stem, irrespective of whether the produced stem constitutes an authentic dictionary word.

#### 14. What is lemmatization?
Lemmatization is a linguistically grounded morphological process that resolves an inflected word form to its canonical, dictionary-valid base form (known as a **lemma**) using lexical databases and part-of-speech context.

#### 15. What is the main difference between stemming and lemmatization?
* **Stemming:** Operates via heuristic pattern-matching and suffix chopping without lexical awareness, often producing non-dictionary word fragments (e.g., `"connection"` $\rightarrow$ `"connect"`, `"studies"` $\rightarrow$ `"studi"`).
* **Lemmatization:** Consults full morphological lexicons (e.g., WordNet) alongside part-of-speech tags to guarantee meaningful, real dictionary entries (e.g., `"studies"` $\rightarrow$ `"study"`, `"better"` $\rightarrow$ `"good"`).

#### 16. Give one example of a word before and after stemming.
* **Input Word:** `operational`
* **After Porter Stemming:** `oper`

#### 17. Give one example of a word before and after lemmatization.
* **Input Word:** `matrices`
* **After WordNet Lemmatization:** `matrix`

#### 18. Which method usually produces meaningful dictionary words: stemming or lemmatization?
**Lemmatization** consistently produces valid dictionary terms, whereas stemming regularly yields truncated stems that are not legitimate linguistic words.

#### 19. Write the correct sequence of the text preprocessing steps covered in this lab.
The standard canonical pipeline sequence is:
1. Sentence & Word Tokenization
2. Case Folding (Lowercasing)
3. Punctuation Filtering
4. Stopword Elimination
5. Stemming / Lemmatization Normalization

#### 20. What problems can occur if too much information is removed during text preprocessing?
Over-cleansing can strip away critical linguistic context:
* Syntactic order and sequence information are destroyed, which impairs transformer and language model performance.
* Negation and nuance words are discarded, destroying sentiment detection.
* Numerical quantities, dates, and casing cues (which distinguish proper nouns like `"Apple"` from `"apple"`) are erased, harming Named Entity Recognition (NER).


# Supplemental Tasks: Statistical Profiling & Comparative Pipeline Analysis


```python
### Task 3: Comprehensive Preprocessing Pipeline & Statistical Frequency Profiling
from collections import Counter

CORPUS_PATH = "sample_text.txt"

def print_section(heading: str, payload):
    print("\n" + "=" * 70)
    print(heading)
    print("=" * 70)
    print(payload)

def dual_pass_lemmatize(word_token: str, lemma_engine) -> str:
    """Applies verbal lemmatization followed by nominal resolution."""
    v_lemma = lemma_engine.lemmatize(word_token, pos="v")
    return lemma_engine.lemmatize(v_lemma, pos="n")

# 1. Document Acquisition
with open(CORPUS_PATH, "r", encoding="utf-8") as f_in:
    raw_corpus = f_in.read()

print_section("1. SOURCE TEXT CORPUS", raw_corpus)

# 2. Sentence Segmentation
sentence_units = sent_tokenize(raw_corpus)
print_section("2. SENTENCE SEGMENTATION",
              "\n".join(f"[{i:02d}] {s}" for i, s in enumerate(sentence_units, 1)))

# 3. Lexical Tokenization
raw_word_tokens = word_tokenize(raw_corpus)
print_section("3. RAW TOKEN STREAM (INCLUDING PUNCTUATION)", raw_word_tokens)

# 4. Lexical Statistics
alphanumeric_words = [t for t in raw_word_tokens if t not in string.punctuation]
distinct_vocabulary = set(w.lower() for w in alphanumeric_words)
lexical_diversity = len(distinct_vocabulary) / len(alphanumeric_words) if alphanumeric_words else 0

print_section("4. LEXICAL DESCRIPTIVE STATISTICS",
              f"Sentence Count        : {len(sentence_units)}\n"
              f"Total Word Tokens     : {len(alphanumeric_words)}\n"
              f"Unique Word Vocabulary: {len(distinct_vocabulary)} (case-normalized)\n"
              f"Lexical Diversity     : {lexical_diversity:.3f}")

# 5. Case Folding
corpus_lowered = raw_corpus.lower()
print_section("5. LOWERCASE NORMALIZED CORPUS", corpus_lowered)

# 6. Punctuation Filtering
clean_str = corpus_lowered.translate(str.maketrans("", "", string.punctuation))
clean_str = " ".join(clean_str.split())
print_section("6. TEXT POST PUNCTUATION STRIPPING", clean_str)

extracted_tokens = word_tokenize(clean_str)

# 7. Stopword Elimination
eng_stop_words = set(stopwords.words("english"))
pruned_stopwords = [t for t in extracted_tokens if t in eng_stop_words]
informative_tokens = [t for t in extracted_tokens if t not in eng_stop_words]

print_section("7. TEXT POST STOPWORD PRUNING", " ".join(informative_tokens))
print_section("8. STOPWORD PRUNING SUMMARY",
              f"Filtered stopword instances : {len(pruned_stopwords)}\n"
              f"Unique stopword vocabulary  : {len(set(pruned_stopwords))}")

# 9. Porter Stemming & Mutation Map
stemmer_inst = PorterStemmer()
stemmed_list = [stemmer_inst.stem(t) for t in informative_tokens]
stem_mutations = sorted(set(zip(informative_tokens, stemmed_list)))
stem_diffs = [(w, s) for w, s in stem_mutations if w != s]

print_section("9. PORTER STEMMING TRANSFORMATION",
              "Stemmed Output:\n" + " ".join(stemmed_list) + "\n\n" +
              f"Mutated Words ({len(stem_diffs)} distinct transformations):\n" +
              "\n".join(f"  {w:<16} -> {s}" for w, s in stem_diffs))

# 10. WordNet Lemmatization & Mutation Map
lemma_inst = WordNetLemmatizer()
lemmatized_list = [dual_pass_lemmatize(t, lemma_inst) for t in informative_tokens]
lemma_mutations = sorted(set(zip(informative_tokens, lemmatized_list)))
lemma_diffs = [(w, l) for w, l in lemma_mutations if w != l]

print_section("10. WORDNET LEMMATIZATION TRANSFORMATION",
              "Lemmatized Output:\n" + " ".join(lemmatized_list) + "\n\n" +
              f"Mutated Words ({len(lemma_diffs)} distinct transformations):\n" +
              "\n".join(f"  {w:<16} -> {l}" for w, l in lemma_diffs))

# 11. Final Cleaned Corpus
final_corpus_tokens = lemmatized_list
final_cleaned_string = " ".join(final_corpus_tokens)
print_section("11. SYNTHESIZED FINAL CLEANED CORPUS", final_cleaned_string)

# 12. Volume Reduction
words_prior = len(alphanumeric_words)
words_post = len(final_corpus_tokens)
words_cut = words_prior - words_post
print_section("12. VOLUME COMPRESSION ANALYSIS",
              f"Token Volume (Initial)   : {words_prior}\n"
              f"Token Volume (Cleaned)   : {words_post}\n"
              f"Net Tokens Removed       : {words_cut} ({(words_cut / words_prior):.1%} reduction)")

# 13. Top-10 Term Frequency Analysis
top_frequencies = Counter(final_corpus_tokens).most_common(10)
print_section("13. TOP 10 HIGH-FREQUENCY TERMS IN CLEANED CORPUS",
              "\n".join(f"  [{rank:02d}] {term:<15} : {freq}" for rank, (term, freq) in enumerate(top_frequencies, 1)))
```

```python
### Task 4: Comparative Evaluation: Stemming Pipeline vs. Lemmatization Pipeline
# Pipeline Alpha: Lowercase -> Tokenize -> Filter Punctuation -> Filter Stopwords -> Porter Stemming
# Pipeline Beta : Lowercase -> Tokenize -> Filter Punctuation -> Filter Stopwords -> WordNet Lemmatization

sample_article = (
    "Artificial intelligence and autonomous systems are revolutionizing modern scientific discovery. "
    "Robotic telescopes continuously observe astronomical phenomena, while automated instruments capture "
    "enormous datasets across global laboratories. Researchers are training deep neural networks to recognize "
    "subtle biological patterns in microscopic images, identifying early indicators of cellular diseases that "
    "human observers might easily overlook. Furthermore, distributed computing frameworks accelerate simulation "
    "experiments, allowing scientists to investigate complex molecular interactions within minutes. Although computational "
    "challenges persist, such as managing storage limitations and processing noisy measurements, these emerging "
    "technologies empower academic communities to collaborate efficiently and advance medical research for humanity."
)

assert len(sample_article.split()) >= 100, f"Sample text must contain at least 100 words (Current: {len(sample_article.split())})"

active_stopwords = set(stopwords.words("english"))
stemming_engine = PorterStemmer()
lemmatizing_engine = WordNetLemmatizer()

def standard_cleanup_pipeline(raw_content: str):
    """Performs initial standardization: lowercasing, tokenization, punctuation and stopword removal."""
    toks = word_tokenize(raw_content.lower())
    # Strip standalone punctuation and punctuation-wrapped boundaries
    toks = [tok.strip(string.punctuation) for tok in toks]
    toks = [tok for tok in toks if tok and tok not in string.punctuation]
    return [tok for tok in toks if tok not in active_stopwords]

def apply_full_lemmatization(word: str) -> str:
    """Resolves verbs and nouns to base morphological lemma."""
    v_lem = lemmatizing_engine.lemmatize(word, pos="v")
    return lemmatizing_engine.lemmatize(v_lem, pos="n")

# Execute pipelines
shared_clean_tokens = standard_cleanup_pipeline(sample_article)
pipeline_alpha = [stemming_engine.stem(t) for t in shared_clean_tokens]      # Stemming
pipeline_beta = [apply_full_lemmatization(t) for t in shared_clean_tokens]    # Lemmatization

print("=" * 70)
print("SOURCE INPUT ARTICLE")
print("=" * 70)
print(sample_article)
print(f"\n(Initial Word Count: {len(sample_article.split())} words)")

print("\n" + "=" * 70)
print("PIPELINE ALPHA (Porter Stemming Result)")
print("=" * 70)
print(" ".join(pipeline_alpha))

print("\n" + "=" * 70)
print("PIPELINE BETA (WordNet Lemmatization Result)")
print("=" * 70)
print(" ".join(pipeline_beta))

print("\n" + "=" * 70)
print("QUANTITATIVE COMPARISON")
print("=" * 70)
print(f"Pipeline Alpha (Stemming)      : {len(pipeline_alpha)} total tokens | {len(set(pipeline_alpha))} unique terms")
print(f"Pipeline Beta  (Lemmatization) : {len(pipeline_beta)} total tokens | {len(set(pipeline_beta))} unique terms")

print("\n" + "=" * 70)
print("TOKEN DIVERGENCE ANALYSIS (Alpha vs Beta)")
print("=" * 70)
print(f"{'Source Token':<18}{'Pipeline Alpha (Stem)':<24}{'Pipeline Beta (Lemma)':<24}")
print("-" * 66)

divergence_records = set()
for src, a_tok, b_tok in zip(shared_clean_tokens, pipeline_alpha, pipeline_beta):
    if a_tok != b_tok and (src, a_tok, b_tok) not in divergence_records:
        divergence_records.add((src, a_tok, b_tok))
        print(f"{src:<18}{a_tok:<24}{b_tok:<24}")

print(f"\nTotal distinct divergent tokens identified: {len(divergence_records)}")

print("\n" + "=" * 70)
print("ANALYTICAL EVALUATION: WHICH PIPELINE PRESERVES GREATER MEANING?")
print("=" * 70)
print("""
1. Linguistic Interpretability:
   Lemmatization (Pipeline Beta) clearly preserves superior semantic fidelity and human readability.
   Because it relies on a curated morphological lexical database (WordNet), every output token remains
   a legitimate, meaningful English dictionary word (e.g., 'revolutionary' -> 'revolutionize', 'cellular' -> 'cellular').

2. Structural Distortion in Stemming:
   Stemming (Pipeline Alpha) uses rule-based heuristic suffix slicing. While fast, it regularly truncates
   stems into pseudo-words (e.g., 'astronomical' -> 'astronom', 'discovery' -> 'discoveri', 'efficiently' -> 'effici').
   These non-words can obscure semantic nuances and cause ungrounded collations.

3. Engineering Recommendation:
   - Use Lemmatization when interpretability, entity recognition, sentiment analysis, or downstream LLM
     prompting is the objective.
   - Use Stemming when low computational latency and raw memory speed take precedence (e.g., high-volume inverted
     index construction in large-scale search engines).
""")
```
