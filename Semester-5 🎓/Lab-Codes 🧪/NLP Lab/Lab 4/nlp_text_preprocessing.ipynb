```python
import string
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download necessary NLTK datasets (including 'punkt_tab')
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)
```

# Task 1 - Basic Text Preprocessing

Write a Python program that takes a sentence from the user and performs:

•	Lowercasing

•	Word tokenization

•	Punctuation removal

•	Stopword removal

•	Porter stemming

•	WordNet lemmatization


```python
def preprocess_text(text):
    print(f"\nOriginal Text: {text}\n")

    # 1. Lowercasing
    lowercased = text.lower()
    print(f"1. Lowercased: {lowercased}")

    # 2. Word Tokenization
    tokens = word_tokenize(lowercased)
    print(f"2. Tokenized: {tokens}")

    # 3. Punctuation Removal
    no_punctuation = [token for token in tokens if token not in string.punctuation]
    print(f"3. Punctuation Removed: {no_punctuation}")

    # 4. Stopword Removal
    stop_words = set(stopwords.words('english'))
    no_stopwords = [token for token in no_punctuation if token not in stop_words]
    print(f"4. Stopwords Removed: {no_stopwords}")

    # 5. Porter Stemming
    stemmer = PorterStemmer()
    stemmed = [stemmer.stem(token) for token in no_stopwords]
    print(f"5. Porter Stemmed: {stemmed}")

    # 6. WordNet Lemmatization
    lemmatizer = WordNetLemmatizer()
    lemmatized = [lemmatizer.lemmatize(token) for token in no_stopwords]
    print(f"6. WordNet Lemmatized: {lemmatized}")

    return lemmatized
```

```python
# Take input from the user
user_sentence = input("Enter a sentence to preprocess: ")
preprocessed_result = preprocess_text(user_sentence)
```

# Task 2 - Complete Preprocessing of a Text File

Create a text file containing at least 100 words. Read it in Python and display the result after each preprocessing stage.

•	Original text and original word count

•	Sentence and word tokenization

•	Lowercasing

•	Punctuation removal

•	Stopword removal

•	Stemming

•	Lemmatization

•	Final cleaned text

•	Word count before and after preprocessing


```python
# 1. Read the file and show original text + word count
with open("sample.txt", "r", encoding="utf-8") as f:
    text = f.read()

original_words = text.split()
print("=" * 60)
print("1. ORIGINAL TEXT")
print("=" * 60)
print(text)
print("\nOriginal word count:", len(original_words))
```

```python
# 2. Sentence and word tokenization
sentences = sent_tokenize(text)
words = word_tokenize(text)
print("\n" + "=" * 60)
print("2. TOKENIZATION")
print("=" * 60)
print("Sentence tokens (", len(sentences), "):")
for i, s in enumerate(sentences, 1):
    print(f"  {i}. {s}")
print("\nWord tokens (", len(words), "):")
print(words)
```

```python
# 3. Lowercasing
lower_words = [w.lower() for w in words]
print("\n" + "=" * 60)
print("3. LOWERCASING")
print("=" * 60)
print(lower_words)
```

```python
# 4. Punctuation removal
no_punct = [w for w in lower_words if w not in string.punctuation]
print("\n" + "=" * 60)
print("4. PUNCTUATION REMOVAL")
print("=" * 60)
print(no_punct)
```

```python
# 5. Stopword removal
stop_words = set(stopwords.words("english"))
no_stop = [w for w in no_punct if w not in stop_words]
print("\n" + "=" * 60)
print("5. STOPWORD REMOVAL")
print("=" * 60)
print(no_stop)
```

```python
# 6. Stemming
stemmer = PorterStemmer()
stemmed = [stemmer.stem(w) for w in no_stop]
print("\n" + "=" * 60)
print("6. STEMMING")
print("=" * 60)
print(stemmed)
```

```python
# 7. Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatizer.lemmatize(w, pos='v') for w in no_stop]
print("\n" + "=" * 60)
print("7. LEMMATIZATION")
print("=" * 60)
print(lemmatized)
```

```python
# 8. Final cleaned text
final_text = " ".join(lemmatized)
print("\n" + "=" * 60)
print("8. FINAL CLEANED TEXT")
print("=" * 60)
print(final_text)
```

```python
# 9. Word count before and after
print("\n" + "=" * 60)
print("9. WORD COUNT COMPARISON")
print("=" * 60)
print("Word count BEFORE preprocessing:", len(original_words))
print("Word count AFTER preprocessing :", len(final_text.split()))
```

### Questions
#### 1. What is text preprocessing in NLP?
Text preprocessing is the process of cleaning, normalizing, and preparing raw text data to make it suitable for analysis and modeling by machine learning or deep learning algorithms. It transforms messy human language into a structured, consistent format.

#### 2. Why is text preprocessing important before training an NLP model?
Raw text is highly unstructured and contains noise (such as inconsistent capitalization, punctuation, and filler words). Preprocessing reduces this noise, standardizes the vocabulary, reduces the dimensionality of the feature space, and helps models learn meaningful patterns more efficiently, leading to better accuracy and performance.

#### 3. What is tokenization?
Tokenization is the process of breaking a continuous sequence of character strings (like a paragraph or sentence) into smaller, meaningful units called **tokens** (such as sentences or words).

#### 4. What is the difference between sentence tokenization and word tokenization?
*   **Sentence Tokenization:** Splits a text into its individual sentences (e.g., using periods, exclamation marks, or question marks as boundaries).
*   **Word Tokenization:** Splits a sentence or text into individual words, symbols, and punctuation marks.

#### 5. What is lowercasing in text preprocessing?
Lowercasing is the process of converting all characters in the text into lowercase format (e.g., "Natural" and "natural" both become "natural").

#### 6. Why do we convert text into lowercase?
We convert text to lowercase to maintain consistency and avoid duplicate entries. Computers treat "Run", "RUN", and "run" as three completely different words. Lowercasing maps them to a single token, reducing vocabulary size and solving sparsity issues.

#### 7. What is punctuation removal?
It is the process of filtering out punctuation marks (such as commas, periods, exclamation points, and brackets) from the tokenized text list.

#### 8. Why is punctuation removed from text in some NLP tasks?
In many bag-of-words or classification tasks, punctuation marks do not carry significant semantic meaning but increase the vocabulary size. Removing them helps focus purely on the textual words.

#### 9. What are stopwords?
Stopwords are highly frequent words in a language (such as "the", "is", "at", "which", and "on") that serve structural grammatical purposes but carry little to no unique semantic information about the subject of the text.

#### 10. Give five examples of common English stopwords.
Examples include: `the`, `is`, `and`, `to`, `in`.

#### 11. Why are stopwords removed during text preprocessing?
They are removed to filter out noise, focus the analysis on content-carrying words, and significantly reduce the dimensionality of the text data for tasks like document classification or search.

#### 12. Why should words such as “not” sometimes be kept instead of removed?
In sentiment analysis or opinion mining, negation words like "not" or "no" completely flip the sentiment of a phrase (e.g., "good" vs. "not good"). Removing them would lead to incorrect classification of sentiment.

#### 13. What is stemming?
Stemming is a crude, rule-based heuristic process that chops off the ends of words (like suffixes) to reduce them to their base or root form (called a stem). It does not take part of speech or linguistic context into account.

#### 14. What is lemmatization?
Lemmatization is a vocabulary and morphological analysis process that aims to return the base dictionary form of a word (called a **lemma**). It takes the context and part of speech (POS) of the word into consideration.

#### 15. What is the main difference between stemming and lemmatization?
*   **Stemming** uses fast, crude suffix-stripping rules and often produces non-dictionary words (e.g., "studying" becomes "studi").
*   **Lemmatization** uses a vocabulary database (like WordNet) and morphological analysis to always produce valid, meaningful dictionary words (e.g., "studying" becomes "study").

#### 16. Give one example of a word before and after stemming.
*   **Word:** `happily`
*   **After Stemming (Porter Stemmer):** `happili`

#### 17. Give one example of a word before and after lemmatization.
*   **Word:** `children`
*   **After Lemmatization:** `child`

#### 18. Which method usually produces meaningful dictionary words: stemming or lemmatization?
**Lemmatization** produces meaningful dictionary words, whereas stemming often results in truncated stems that are not real dictionary entries.

#### 20. What problems can occur if too much information is removed during text preprocessing?
Over-cleaning can strip away crucial contextual, grammatical, or semantic cues. This can lead to:
*   Loss of syntax and sentence structure (important for language models and translation).
*   Loss of negations (ruining sentiment analysis).
*   Loss of proper nouns or specific terminology, making it difficult for models to understand who or what is being discussed.

```python
### Task 3 - Preprocessing and Statistical Analysis of a Text File
# Using standard libraries and imports defined at the beginning of the notebook
from collections import Counter

FILE_PATH = "sample_text.txt"

def show(title, content):
    print("\n" + "=" * 70)
    print(title)
    print("=" * 70)
    print(content)

def lemmatize(word, lemmatizer):
    """Try verb first (running -> run), then fall back to noun (diseases -> disease)."""
    lemma = lemmatizer.lemmatize(word, pos="v")
    return lemmatizer.lemmatize(lemma, pos="n")

# ---------------------------------------------------------------- load
with open(FILE_PATH, "r", encoding="utf-8") as f:
    original_text = f.read()

show("1. ORIGINAL TEXT", original_text)

# ---------------------------------------------------------------- tokenization
sentences = sent_tokenize(original_text)
show("2. SENTENCE TOKENIZATION",
     "\n".join(f"[{i}] {s}" for i, s in enumerate(sentences, 1)))

word_tokens = word_tokenize(original_text)
show("3. WORD TOKENIZATION (punctuation marks are separate tokens)", word_tokens)

# ---------------------------------------------------------------- basic statistics
# "Words" = tokens that are not pure punctuation
words_only = [t for t in word_tokens if t not in string.punctuation]
unique_words = set(w.lower() for w in words_only)

show("4. BASIC STATISTICS",
     f"Total number of sentences : {len(sentences)}\n"
     f"Total number of words     : {len(words_only)}\n"
     f"Number of unique words    : {len(unique_words)}  (case-insensitive)")

# ---------------------------------------------------------------- lowercasing
lowered = original_text.lower()
show("5. LOWERCASED TEXT", lowered)

# ---------------------------------------------------------------- punctuation removal
no_punct_text = lowered.translate(str.maketrans("", "", string.punctuation))
no_punct_text = " ".join(no_punct_text.split())      # tidy extra whitespace
show("6. TEXT AFTER PUNCTUATION REMOVAL", no_punct_text)

tokens = word_tokenize(no_punct_text)

# ---------------------------------------------------------------- stopword removal
stop_set = set(stopwords.words("english"))
removed_stopwords = [t for t in tokens if t in stop_set]
no_stop_tokens = [t for t in tokens if t not in stop_set]

show("7. TEXT AFTER STOPWORD REMOVAL", " ".join(no_stop_tokens))
show("8. TOTAL STOPWORDS REMOVED",
     f"{len(removed_stopwords)} stopword tokens removed "
     f"({len(set(removed_stopwords))} distinct stopwords)")

# ---------------------------------------------------------------- stemming
stemmer = PorterStemmer()
stemmed = [stemmer.stem(t) for t in no_stop_tokens]
show("9. STEMMING RESULTS (Porter)",
     "Stemmed text: " + " ".join(stemmed) + "\n\nWord -> Stem (changed words only):\n" +
     "\n".join(f"  {w:<15} -> {s}" for w, s in sorted(set(zip(no_stop_tokens, stemmed))) if w != s))

# ---------------------------------------------------------------- lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatize(t, lemmatizer) for t in no_stop_tokens]
show("10. LEMMATIZATION RESULTS (WordNet)",
     "Lemmatized text: " + " ".join(lemmatized) + "\n\nWord -> Lemma (changed words only):\n" +
     "\n".join(f"  {w:<15} -> {l}" for w, l in sorted(set(zip(no_stop_tokens, lemmatized))) if w != l))

# ---------------------------------------------------------------- final cleaned text
# Lemmatized tokens are used as the final text because they remain real words.
final_tokens = lemmatized
final_text = " ".join(final_tokens)
show("11. FINAL CLEANED TEXT (lowercase, no punctuation, no stopwords, lemmatized)", final_text)

# ---------------------------------------------------------------- before / after counts
show("12. WORD COUNT BEFORE AND AFTER PREPROCESSING",
     f"Before preprocessing : {len(words_only)} words\n"
     f"After preprocessing  : {len(final_tokens)} words\n"
     f"Reduction            : {len(words_only) - len(final_tokens)} words "
     f"({(len(words_only) - len(final_tokens)) / len(words_only):.1%})")

# ---------------------------------------------------------------- top-10 frequent words
top10 = Counter(final_tokens).most_common(10)
show("13. 10 MOST FREQUENT WORDS IN FINAL CLEANED TEXT",
     "\n".join(f"{rank:>2}. {word:<15} {count}" for rank, (word, count) in enumerate(top10, 1)))
```

```python
### Task 4 - Compare Two Preprocessing Approaches
# Version A: Lowercase -> Tokenize -> Remove punctuation -> Remove stopwords -> Stemming
# Version B: Lowercase -> Tokenize -> Remove punctuation -> Remove stopwords -> Lemmatization

paragraph = (
    "Renewable energy is growing faster than many experts predicted. Solar panels are "
    "becoming cheaper every year, and wind turbines are producing electricity in places "
    "where nobody expected them to work. Governments are investing in new technologies, "
    "while companies are studying better ways of storing power for cloudy and windless "
    "days. Communities that once depended on coal are now learning new skills and "
    "building cleaner industries. Although challenges remain, such as expensive batteries "
    "and old power grids, the progress made during the last decade shows that a "
    "sustainable future is possible. Many families are already saving money by installing "
    "panels on their roofs, and children are growing up with a better understanding of "
    "how important clean energy is for the planet."
)

assert len(paragraph.split()) >= 100, "Paragraph must have at least 100 words"

stop_set = set(stopwords.words("english"))
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()


def base_pipeline(text):
    """Lowercase -> tokenize -> remove punctuation -> remove stopwords (shared steps)."""
    tokens = word_tokenize(text.lower())
    tokens = [t.strip(string.punctuation) for t in tokens]
    tokens = [t for t in tokens if t and t not in string.punctuation]
    return [t for t in tokens if t not in stop_set]


def lemmatize(word):
    return lemmatizer.lemmatize(lemmatizer.lemmatize(word, pos="v"), pos="n")


base_tokens = base_pipeline(paragraph)
version_a = [stemmer.stem(t) for t in base_tokens]      # stemming
version_b = [lemmatize(t) for t in base_tokens]          # lemmatization

print("ORIGINAL PARAGRAPH")
print("-" * 70)
print(paragraph)
print(f"\n(Original word count: {len(paragraph.split())})")

print("\nVERSION A  (Stemming)")
print("-" * 70)
print(" ".join(version_a))

print("\nVERSION B  (Lemmatization)")
print("-" * 70)
print(" ".join(version_b))

print("\nFINAL WORD COUNTS")
print("-" * 70)
print(f"Version A (stemming)      : {len(version_a)} words, {len(set(version_a))} unique")
print(f"Version B (lemmatization) : {len(version_b)} words, {len(set(version_b))} unique")

print("\nWORDS THAT DIFFER BETWEEN THE TWO VERSIONS")
print("-" * 70)
print(f"{'Original':<16}{'Version A (stem)':<20}{'Version B (lemma)':<20}")
seen = set()
for orig, a, b in zip(base_tokens, version_a, version_b):
    if a != b and (orig, a, b) not in seen:
        seen.add((orig, a, b))
        print(f"{orig:<16}{a:<20}{b:<20}")
print(f"\nTotal distinct differing words: {len(seen)}")

print("\nWHICH APPROACH PRESERVES MORE MEANINGFUL WORDS?")
print("-" * 70)
print("Lemmatization (Version B) preserves more meaningful words. Both versions keep the")
print("same number of tokens, but stemming chops words by rule, producing non-words such")
print("as 'energi' or 'technolog' that are hard to read and can merge unrelated words.")
print("Lemmatization uses a dictionary (WordNet) to return real base forms such as")
print("'energy' or 'panel', so the output stays readable and keeps its meaning.")
print("Stemming is faster and can help search or indexing, but loses interpretability.")
```
