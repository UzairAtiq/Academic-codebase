```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/Users/uzair/academic-codebase📘/Semester-5 🎓/Lab-Codes 🧪/ML Lab/Lab 1/Datasets/Cleaned_data_for_model.csv', index_col=0)
df = df.sample(n=500, random_state=42)
```

```python
df.head()
```

**Categorical Data**

*property_type*

```python
sns.countplot(x='property_type', data=df)
plt.xticks(rotation=45)
```

```python
df['property_type'].value_counts()
```

```python
counts = df['property_type'].value_counts()

fig, ax = plt.subplots(figsize=(10, 7))

wedges, texts, autotexts = ax.pie(
    counts,
    labels=None,
    autopct='%.2f%%',
    pctdistance=0.7
)

ax.legend(
    wedges,
    counts.index,
    title='Property Type',
    loc='center left',
    bbox_to_anchor=(1, 0, 0.5, 1)
)
plt.show()
```

*location*
